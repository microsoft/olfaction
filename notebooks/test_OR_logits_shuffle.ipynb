```python
from torch.utils.data import DataLoader

from data.m2or import M2OR_Pairs
from dgllife.utils import SMILESToBigraph
from dgllife.utils import CanonicalAtomFeaturizer
from dgllife.utils import SMILESToBigraph

smiles_to_g = SMILESToBigraph(add_self_loop=True, node_featurizer=CanonicalAtomFeaturizer(),
                                edge_featurizer=None)

dataset = M2OR_Pairs(smiles_to_graph=smiles_to_g, 
        n_jobs=1, load=True)
```

```python
from dgllife.utils import ScaffoldSplitter, RandomSplitter

train_set, val_set, test_set = RandomSplitter.train_val_test_split(
    dataset, frac_train=0.8, frac_val=0.1, frac_test=0.1, random_state = 42)

```

```python
import pandas as pd
m2or_df = pd.read_csv('data/datasets/M2OR_original_mol_OR_pairs.csv', sep = ';')
len(m2or_df)
```

```python
m2or_df
```

```python
# have columns only be 'smiles','sequences','label'
perceiver_m2or_df = pd.DataFrame(columns=['smiles','sequence','label'])

perceiver_m2or_df['smiles'] = m2or_df['canonicalSMILES']
perceiver_m2or_df['sequence'] = m2or_df['mutated_Sequence']
perceiver_m2or_df['label'] = m2or_df['Responsive']
```

```python
perceiver_m2or_df.head()
```

```python
# grab train indiices from m2or_df using train_set.indices, an array of indices
train_df = perceiver_m2or_df.iloc[train_set.indices]
val_df = perceiver_m2or_df.iloc[val_set.indices]
test_df = perceiver_m2or_df.iloc[test_set.indices]
```
