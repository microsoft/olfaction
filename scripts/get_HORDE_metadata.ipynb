```python
import pandas as pd
mol_OR = pd.read_csv('../data/datasets/genes.csv', delimiter='\t')
# Sequence order used for logits: value_counts over Conceptual Sequence (no cleaning)
top_seqs = mol_OR['Conceptual Sequence'].value_counts().keys().tolist()
```

```python
mol_OR['Pseduo Gene'].value_counts()
```

```python
#number of unique sequences in HORDE
mol_OR['Conceptual Sequence'].nunique()
```

```python
len(top_seqs)
```

```python
# Build per-sequence pseudogene flag and indices (no cleaning)
# Assumes identical Pseduo Gene value per Conceptual Sequence
flag_by_seq = mol_OR.drop_duplicates(subset=['Conceptual Sequence']).set_index('Conceptual Sequence')['Pseduo Gene'].astype(int)

pseudogene_indices = [i for i, s in enumerate(top_seqs) if flag_by_seq.get(s, 0) == 1]
functional_indices = [i for i, s in enumerate(top_seqs) if flag_by_seq.get(s, 0) == 0]

len(top_seqs), len(pseudogene_indices), len(functional_indices)
```

```python
# Save to data/datasets/HORDE/
import json, os
out_dir = os.path.join('..', 'data', 'datasets', 'HORDE')
os.makedirs(out_dir, exist_ok=True)
with open(os.path.join(out_dir, 'pseudogene_indices.json'), 'w') as f:
    json.dump(pseudogene_indices, f)
with open(os.path.join(out_dir, 'functional_indices.json'), 'w') as f:
    json.dump(functional_indices, f)
print('Saved:', out_dir)
```
